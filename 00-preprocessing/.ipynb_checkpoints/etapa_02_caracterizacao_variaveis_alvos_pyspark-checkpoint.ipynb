**Bases:** `df01` SINAN-Dengue (2024), `df02` Doença Renal Crônica, `df03` MIMIC-II IACCD, `df04` Terapia Periodontal, `df05` NHANES.

**Objetivos:** inspecionar dimensionalidade, nomes, tipos técnicos e semânticos, cardinalidades, proporções de ausências e distribuição das classes; gerar relatórios para decidir as variáveis-alvo e os preditores.

**Não modifica nem apaga os CSVs originais.** Nesta etapa nenhuma coluna é eliminada automaticamente. As sugestões de alvo e de tipo de atributo precisam de revisão humana.

## 0. Preparação
Use um kernel Python que tenha `pyspark`, `pandas` e `ipykernel`. Configure a memória de acordo com o seu equipamento. Para o SINAN, a leitura é distribuída; apenas os resumos pequenos são convertidos em pandas.

```python
from pathlib import Path
import csv, re, json, unicodedata
import pandas as pd
from IPython.display import display
from pyspark.sql import SparkSession, functions as F

BASE_DIR = Path('/home/mirlei/datasets_rl')
RAW_DIR = BASE_DIR / '0_raw'
REPORT_DIR = BASE_DIR / 'reports' / 'etapa_02'
REPORT_DIR.mkdir(parents=True, exist_ok=True)

DATASETS = {
    'df01': 'SINAN Dengue 2024',
    'df02': 'Chronic kidney disease',
    'df03': 'MIMIC-II arterial',
    'df04': 'Periodontal therapy',
    'df05': 'NHANES Age Prediction',
}
# Correções manuais caso algum CSV use separador/encoding especial:
SEPARADORES = {}  # ex.: {'df01': ';'}
ENCODINGS = {}  # ex.: {'df01': 'ISO-8859-1'}

spark = (SparkSession.builder.appName('datasets_rl_etapa_02')
         .master('local[*]')
         .config('spark.driver.memory', '6g')
         .config('spark.sql.shuffle.partitions', '16')
         .getOrCreate())
spark.sparkContext.setLogLevel('WARN')
print('Spark', spark.version, '| Origem:', RAW_DIR, '| Relatórios:', REPORT_DIR)
```

## 1. Leitura segura
Mantemos **todos os campos como string** na entrada para preservar categorias codificadas (inclusive `01`, `02` etc.) e não interpretar acidentalmente códigos como valores numéricos. Uma coluna quantitativa será identificada depois, por diagnóstico. A detecção automática de delimitador é conferida na tela.

```python
def detectar_sep(caminho):
    with caminho.open('r', encoding='utf-8-sig', errors='replace', newline='') as arq:
        amostra = arq.read(65536)
    try:
        return csv.Sniffer().sniff(amostra, delimiters=',;\t|').delimiter
    except csv.Error:
        cab = amostra.splitlines()[0] if amostra else ''
        return max([',', ';', '\t', '|'], key=lambda sep: cab.count(sep))

config = {}
for id_, nome in DATASETS.items():
    caminho = RAW_DIR / f'{id_}.csv'
    if not caminho.exists():
        print('ARQUIVO NÃO ENCONTRADO:', caminho)
        continue
    sep = SEPARADORES.get(id_, detectar_sep(caminho))
    config[id_] = {'path': caminho, 'sep': sep, 'encoding': ENCODINGS.get(id_, 'UTF-8')}
    print(f'{id_}: {nome} | {caminho.stat().st_size / (1024**2):,.1f} MiB | separador={repr(sep)}')
assert config, 'Nenhum CSV encontrado. Verifique RAW_DIR.'

def carregar(id_):
    opt = config[id_]
    df = (spark.read.option('header', True).option('sep', opt['sep'])
          .option('encoding', opt['encoding']).option('inferSchema', False)
          .option('mode', 'FAILFAST').csv(str(opt['path'])))
    if len(df.columns) != len(set(df.columns)):
        raise ValueError(f'{id_}: existem colunas com nomes duplicados.')
    if len(df.columns) <= 1:
        raise ValueError(f'{id_}: foi lida apenas 1 coluna; reveja o separador {opt["sep"]!r}.')
    return df

for id_ in config:
    d = carregar(id_)
    print(f'{id_}: {len(d.columns)} colunas | amostra de nomes: {d.columns[:8]}')
```

## 2. Sugestões de variável-alvo (NÃO confirmadas)

As sugestões procuram nomes de coluna. A definição depende do **dicionário oficial e do significado do rótulo**. No SINAN, `CLASSI_FIN` pode ter diferentes codificações/documentações; verifique os valores reais antes de escolher gravidade ou confirmação. No NHANES, o alvo de faixa etária deriva da idade, portanto **não inclua idade nos preditores**.

Outros cuidados: não incluir campos de diagnóstico, critérios de classificação, desfecho ou eventos posteriores à predição se revelarem a classe.

```python
# Ajuste livremente os nomes após conferir as colunas do CSV.
ALVOS = {
    'df01': None,   # provável CLASSI_FIN; CONFIRMAR valores e dicionário
    'df02': None,   # procurar class / classification / target
    'df03': None,   # procurar mortalidade em 28 dias ou equivalente
    'df04': None,   # provável diagnosis, mas avaliar vazamento
    'df05': None,   # procurar age_group / age_class; excluir idade dos preditores
}
PADROES_ALVO = [
    'classi_fin', 'class', 'classification', 'target', 'label', 'diagnosis',
    'outcome', 'mortality', 'death', 'died', 'death28', '28day',
    'age_group', 'agegrp', 'age_class', 'risk', 'result', 'status', 'evolucao'
]

def normalizar_nome(nome):
    return unicodedata.normalize('NFKD', nome).encode('ascii', 'ignore').decode().lower()

def sugestoes_alvo(colunas):
    return [c for c in colunas if any(k in normalizar_nome(c) for k in PADROES_ALVO)]

for id_ in config:
    cols = carregar(id_).columns
    print(id_, '| candidatas por nome:', sugestoes_alvo(cols) or 'nenhuma — revisar cabeçalho')
    print('  nomes completos:', cols)
```

## 3. Relatório geral das cinco bases
Conta linhas e colunas. A contagem é uma operação completa sobre cada arquivo e pode levar alguns minutos no SINAN.

```python
resumo = []
for id_ in config:
    d = carregar(id_)
    n = d.count()
    resumo.append({'dataset': id_, 'nome': DATASETS[id_], 'registros': n,
                   'colunas': len(d.columns), 'alvo_configurado': ALVOS[id_] or ''})
resumo_pd = pd.DataFrame(resumo)
display(resumo_pd)
resumo_pd.to_csv(REPORT_DIR / 'dimensoes.csv', index=False)
```

## 4. Perfil por atributo (tipos, cardinalidade e nulos)

* `nulos`: `NULL` do Spark, strings vazias e marcadores textuais comuns (`NA`, `N/A`, `NULL`, `NONE`, `NAN`, `?`). **Códigos como `9`, `99` e `999` NÃO são convertidos automaticamente** porque dependem do dicionário e podem ser valores válidos em outra coluna.
* `distintos_aprox`: quantidade **aproximada** de valores diferentes, útil para milhões de registros; não é adequada para provar cardinalidade exata.
* `tipo_sugerido`: regra exploratória baseada no nome, na cardinalidade aproximada e no percentual de valores aparentemente numéricos de uma pequena amostra. **Valide manualmente** sobretudo campos codificados como números, datas, identificadores, sexo e faixas etárias.
* A amostra de até 5.000 linhas serve apenas para sugerir tipos; **não** é uma amostra aleatória representativa.

Para evitar uma consulta imensa, as agregações são feitas em lotes de colunas. O SINAN pode exigir processamento demorado, mas não é carregado inteiro em pandas.

```python
MARCADORES = {'', 'null', 'none', 'nan', 'na', 'n/a', '?'}
TAM_LOTE = 16
N_AMOSTRA_TIPO = 5000

# Aspas por segurança em nomes contendo ponto/espaço/backtick
def col(nome):
    return F.col('`' + nome.replace('`', '``') + '`')

def vazio(nome):
    v = F.lower(F.trim(col(nome)))
    return col(nome).isNull() | v.isin(*sorted(MARCADORES))

def numerico_simples(s):
    s = str(s).strip().replace(',', '.')
    return bool(re.fullmatch(r'[+-]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][+-]?\d+)?', s))

def sugestao_tipo(nome, n_dist, n_validos, numericos_prop):
    nm = normalizar_nome(nome)
    if re.search(r'(^id_|_id$|municip|cnes|cep|codigo|cod_|cpf|cns|unidade|_uf$|^sg_uf|^co_)', nm):
        return 'identificador/codigo (revisar)'
    if re.search(r'(^dt_|_date|date_|data_|^sem_|^ano_)', nm):
        return 'data/tempo (revisar)'
    if n_validos == 0:
        return 'sem valores validos'
    if n_dist <= 2:
        return 'categorico binario (revisar)'
    if n_dist <= 15:
        return 'categorico/ordinal (revisar)'
    if numericos_prop >= 0.98:
        return 'numerico provavel (revisar se codigo)'
    return 'texto/categorico alta cardinalidade'

def perfil_base(id_):
    d = carregar(id_)
    n = next(r['registros'] for r in resumo if r['dataset'] == id_)
    nomes = d.columns
    # Amostra pequena apenas para avaliar parseabilidade numérica
    amostra = d.limit(N_AMOSTRA_TIPO).toPandas()
    linhas = []
    for pos in range(0, len(nomes), TAM_LOTE):
        lote = nomes[pos:pos+TAM_LOTE]
        expressions = []
        for j, nome in enumerate(lote):
            expressions += [
                F.sum(F.when(vazio(nome), 1).otherwise(0)).alias(f'm_{j}'),
                F.approx_count_distinct(F.when(~vazio(nome), col(nome)), rsd=0.05).alias(f'd_{j}')
            ]
        val = d.agg(*expressions).first().asDict()
        for j, nome in enumerate(lote):
            aus = int(val[f'm_{j}'] or 0)
            distintos = int(val[f'd_{j}'] or 0)
            serie = amostra[nome].dropna().astype(str).str.strip()
            serie = serie[~serie.str.lower().isin(MARCADORES)]
            numericos_prop = (float(serie.map(numerico_simples).mean()) if len(serie) else 0.0)
            validos = n-aus
            linhas.append({'dataset': id_, 'coluna': nome, 'tipo_spark': 'string',
                           'registros': n, 'nulos': aus, 'pct_nulos': round(100*aus/n,3) if n else 0,
                           'validos': validos, 'distintos_aprox': distintos,
                           'pct_numericos_amostra': round(100*numericos_prop,1),
                           'tipo_sugerido': sugestao_tipo(nome, distintos, validos, numericos_prop)})
    p = pd.DataFrame(linhas).sort_values(['pct_nulos', 'coluna'], ascending=[False, True])
    p.to_csv(REPORT_DIR / f'{id_}_perfil_colunas.csv', index=False)
    return p

PERFIS = {}
for id_ in config:
    print('Caracterizando', id_, DATASETS[id_])
    PERFIS[id_] = perfil_base(id_)
    print('   concluído:', len(PERFIS[id_]), 'atributos')
print('Relatórios em:', REPORT_DIR)
```

## 5. Explorar o perfil e editar os tipos semânticos
A escolha de `numérico`, `nominal`, `ordinal`, `binário`, `data`, `identificador`, `alvo` ou `excluir` deve ser **manual e justificada** depois da inspeção do dicionário. Esta célula gera uma planilha CSV para você editar em LibreOffice, Excel ou pandas.

```python
BASE_INSPECAO = 'df01'  # altere de df01 a df05
display(PERFIS[BASE_INSPECAO][['coluna','pct_nulos','distintos_aprox',
                                'pct_numericos_amostra','tipo_sugerido']].reset_index(drop=True))

for id_, p in PERFIS.items():
    fichas = p[['dataset','coluna','tipo_sugerido']].copy()
    fichas['tipo_validado'] = ''  # numérico | nominal | ordinal | binário | data | identificador | alvo | excluir
    fichas['justificativa'] = ''
    fichas.to_csv(REPORT_DIR / f'{id_}_revisao_tipos.csv', index=False)
print('Arquivos *_revisao_tipos.csv gerados para revisão manual.')
```

## 6. Distribuição das variáveis candidatas a classe

Primeiro, ajuste `ALVOS` na seção 2 e execute **esta célula** novamente. O código mostra distribuição exata por valor, incluindo nulos textuais. Uma classe com alta cardinalidade, como identificadores, será bloqueada pela proteção abaixo.

**ATENÇÃO SINAN:** o dicionário fornecido chama `classi_fin` de classificação final (confirmado/descartado/inconclusivo), mas também menciona valores 11/12 para gravidade. Verifique a codificação **real** do arquivo de 2024 antes de fixar seu objetivo experimental. Para classificar a gravidade, também avalie elegibilidade e vazamento por sinais de alarme/gravidade registrados após o desfecho.

```python
# Configure nomes conforme os cabeçalhos vistos na seção 2.
# Exemplo, depois de confirmar: ALVOS['df01'] = 'CLASSI_FIN'
# ALVOS['df04'] = 'diagnosis'

DISTRIBUICOES = {}
for id_ in config:
    d = carregar(id_)
    alvo = ALVOS[id_]
    if not alvo:
        print(f'{id_}: alvo ainda não definido. Candidatas: {sugestoes_alvo(d.columns)}')
        continue
    correspondencias = [c for c in d.columns if c.lower() == alvo.lower()]
    if len(correspondencias) != 1:
        print(f'{id_}: coluna alvo {alvo!r} não encontrada ou ambígua.')
        continue
    real = correspondencias[0]
    dist = (d.withColumn('_classe', F.when(vazio(real), F.lit('<AUSENTE>')).otherwise(F.trim(col(real))))
            .groupBy('_classe').count().orderBy(F.desc('count')))
    k = dist.count()
    if k > 100:
        print(f'{id_}: {k} valores distintos. Pode não ser uma classe; NÃO converter distribuição inteira para pandas.')
        dist.show(20, truncate=False)
        continue
    pdf = dist.toPandas().rename(columns={'_classe':'classe'})
    pdf['pct'] = (100 * pdf['count'] / pdf['count'].sum()).round(3)
    pdf.insert(0, 'dataset', id_)
    pdf.insert(1, 'coluna_alvo', real)
    display(pdf)
    pdf.to_csv(REPORT_DIR / f'{id_}_distribuicao_alvo.csv', index=False)
    DISTRIBUICOES[id_] = pdf
```

## 7. Auditar códigos de 'ignorado' e categorias condicionais
Códigos especiais como `9`, `99`, `0`, `8`, `Não se aplica` não são ausentes em todas as colunas. A célula lista frequências de **uma coluna por vez**, para análise com o dicionário. No SINAN, campos relativos a internação, laboratório e gravidade podem ficar legitimamente em branco por não serem aplicáveis.

```python
BASE_INSPECAO = 'df01'
COLUNA_INSPECAO = None  # ex.: 'CS_GESTANT', 'CLASSI_FIN', 'CS_RACA'
if COLUNA_INSPECAO:
    d = carregar(BASE_INSPECAO)
    real = next((c for c in d.columns if c.lower()==COLUNA_INSPECAO.lower()), None)
    if not real:
        raise ValueError('Coluna não encontrada: ' + COLUNA_INSPECAO)
    d.groupBy(col(real)).count().orderBy(F.desc('count')).show(50, truncate=False)
else:
    print('Defina COLUNA_INSPECAO para ver valores e códigos (sem substituí-los automaticamente).')
```

## 8. Planejar seleção de atributos e evitar vazamento

Recomenda-se documentar por base:
- alvo e **significado clínico das classes**;
- momento da predição (dados disponíveis na notificação, admissão etc.);
- colunas a excluir: identificadores, datas posteriores, variáveis que codificam ou são consequência do desfecho;
- tipos de cada preditor após validação;
- número de registros por classe **depois** de aplicar critérios de inclusão e remover ausentes.

Não aplique automaticamente a regra de 50% ao **alvo**. Uma base em que a classe tem grande proporção de nulos não pode perder a coluna-alvo e continuar válida para classificação. Primeiro filtre os registros com rótulos válidos e examine os percentuais de nulos **nos preditores elegíveis**, depois decida quais remover.

```python
PLANO = {
    id_: {'alvo': ALVOS[id_], 'classes_validas': [], 'excluir': [],
          'momento_predicao': '', 'notas': ''}
    for id_ in config
}
# Preencha após inspecionar as distribuições e os dicionários, por exemplo:
# PLANO['df01']['alvo'] = 'CLASSI_FIN'
# PLANO['df01']['classes_validas'] = ['10', '11', '12']  # APENAS SE ESSAS CLASSES FOREM CONFIRMADAS NO CSV
# PLANO['df01']['excluir'] = ['EVOLUCAO', 'DT_OBITO', ...]  # verificar nomes reais e temporalidade

arquivo_plano = REPORT_DIR / 'plano_variaveis_alvo.json'
with arquivo_plano.open('w', encoding='utf-8') as f:
    json.dump(PLANO, f, ensure_ascii=False, indent=2)
print('Plano registrado em:', arquivo_plano)

# Resumo dos relatórios gerados
print('\n'.join(str(p.name) for p in sorted(REPORT_DIR.glob('*'))))
```
